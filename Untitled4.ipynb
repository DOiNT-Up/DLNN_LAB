```python
# ============================================================
# TRANSFER LEARNING USING MOBILENETV2 (FAST TESTING VERSION)
# Binary Image Classification: Cats vs Dogs
# ============================================================

# ------------------------------------------------------------
# 1. Import libraries
# ------------------------------------------------------------

import tensorflow as tf
import os
import matplotlib.pyplot as plt

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input


# ------------------------------------------------------------
# 2. Download Cats vs Dogs dataset
# ------------------------------------------------------------

!wget -q https://download.microsoft.com/download/3/E/1/3E1C3F21-ECDB-4869-8368-6DEBA77B919F/kagglecatsanddogs_5340.zip
!unzip -q kagglecatsanddogs_5340.zip


# ------------------------------------------------------------
# 3. Remove corrupted images
# ------------------------------------------------------------

num_skipped = 0

for folder_name in ("Cat", "Dog"):
    folder_path = os.path.join("PetImages", folder_name)
    for fname in os.listdir(folder_path):
        fpath = os.path.join(folder_path, fname)
        try:
            with open(fpath, "rb") as fobj:
                is_jfif = b"JFIF" in fobj.peek(10)
        except Exception:
            is_jfif = False

        if not is_jfif:
            num_skipped += 1
            os.remove(fpath)

print("Deleted corrupted images:", num_skipped)


# ------------------------------------------------------------
# 4. Load dataset and preserve class names
# ------------------------------------------------------------

train_dataset = tf.keras.utils.image_dataset_from_directory(
    "PetImages",
    validation_split=0.2,
    subset="training",
    seed=42,
    image_size=(224, 224),
    batch_size=32
)

validation_dataset = tf.keras.utils.image_dataset_from_directory(
    "PetImages",
    validation_split=0.2,
    subset="validation",
    seed=42,
    image_size=(224, 224),
    batch_size=32
)

# Print class names safely before reducing dataset size
print("Classes:", train_dataset.class_names)

# Limit dataset batches for instant code execution/testing
train_dataset = train_dataset.take(5)       # ~160 images
validation_dataset = validation_dataset.take(2) # ~64 images


# ------------------------------------------------------------
# 5. Preprocess images for MobileNetV2
# ------------------------------------------------------------

train_dataset = train_dataset.map(
    lambda x, y: (preprocess_input(x), y)
)

validation_dataset = validation_dataset.map(
    lambda x, y: (preprocess_input(x), y)
)


# ------------------------------------------------------------
# 6. Load pre-trained MobileNetV2
# ------------------------------------------------------------

base_model = MobileNetV2(
    weights="imagenet",
    include_top=False,
    input_shape=(224, 224, 3)
)


# ------------------------------------------------------------
# 7. Freeze the pre-trained layers
# ------------------------------------------------------------

base_model.trainable = False


# ------------------------------------------------------------
# 8. Add custom classification head
# ------------------------------------------------------------

model = Sequential([
    base_model,
    GlobalAveragePooling2D(),
    Dense(128, activation="relu"),
    Dense(1, activation="sigmoid")
])


# ------------------------------------------------------------
# 9. Compile model
# ------------------------------------------------------------

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)


# ------------------------------------------------------------
# 10. Display model architecture
# ------------------------------------------------------------

model.summary()


# ------------------------------------------------------------
# 11. Train classification head
# ------------------------------------------------------------

history = model.fit(
    train_dataset,
    validation_data=validation_dataset,
    epochs=3
)


# ------------------------------------------------------------
# 12. Evaluate before fine-tuning
# ------------------------------------------------------------

loss, accuracy = model.evaluate(validation_dataset)
print("Accuracy before fine-tuning:", accuracy)


# ============================================================
# FINE-TUNING
# ============================================================


# ------------------------------------------------------------
# 13. Unfreeze the MobileNetV2
# ------------------------------------------------------------

base_model.trainable = True


# ------------------------------------------------------------
# 14. Freeze most layers and fine-tune last few layers
# ------------------------------------------------------------

for layer in base_model.layers[:-20]:
    layer.trainable = False


# ------------------------------------------------------------
# 15. Recompile with smaller learning rate
# ------------------------------------------------------------

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.00001),
    loss="binary_crossentropy",
    metrics=["accuracy"]
)


# ------------------------------------------------------------
# 16. Fine-tune model
# ------------------------------------------------------------

history_fine = model.fit(
    train_dataset,
    validation_data=validation_dataset,
    epochs=3
)


# ------------------------------------------------------------
# 17. Final evaluation
# ------------------------------------------------------------

loss, accuracy = model.evaluate(validation_dataset)
print("Final Test Loss:", loss)
print("Final Test Accuracy:", accuracy)


# ------------------------------------------------------------
# 18. Plot accuracy
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))
plt.plot(history.history["accuracy"], label="Initial Training")
plt.plot(history.history["val_accuracy"], label="Initial Validation")

plt.plot(
    range(
        len(history.history["accuracy"]),
        len(history.history["accuracy"]) + len(history_fine.history["accuracy"])
    ),
    history_fine.history["accuracy"],
    label="Fine-tuning Training"
)

plt.plot(
    range(
        len(history.history["val_accuracy"]),
        len(history.history["val_accuracy"]) + len(history_fine.history["val_accuracy"])
    ),
    history_fine.history["val_accuracy"],
    label="Fine-tuning Validation"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Transfer Learning and Fine-Tuning")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# ============================================================
# WORKFLOW: MANUAL CNN -> TRANSFER LEARNING -> ENSEMBLE COMBINATION
# ============================================================

import tensorflow as tf
import os
import numpy as np
import matplotlib.pyplot as plt

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout, GlobalAveragePooling2D
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input

# ------------------------------------------------------------
# 1. Download & Prepare Dataset (Fast Testing Subset)
# ------------------------------------------------------------
!wget -q https://download.microsoft.com/download/3/E/1/3E1C3F21-ECDB-4869-8368-6DEBA77B919F/kagglecatsanddogs_5340.zip
!unzip -q kagglecatsanddogs_5340.zip

# Clean corrupted images
for folder_name in ("Cat", "Dog"):
    folder_path = os.path.join("PetImages", folder_name)
    for fname in os.listdir(folder_path):
        fpath = os.path.join(folder_path, fname)
        try:
            with open(fpath, "rb") as fobj:
                is_jfif = b"JFIF" in fobj.peek(10)
        except Exception:
            is_jfif = False
        if not is_jfif:
            os.remove(fpath)

# Load data
train_dataset = tf.keras.utils.image_dataset_from_directory(
    "PetImages", validation_split=0.2, subset="training", seed=42, image_size=(150, 150), batch_size=32
)
validation_dataset = tf.keras.utils.image_dataset_from_directory(
    "PetImages", validation_split=0.2, subset="validation", seed=42, image_size=(150, 150), batch_size=32
)

print("Classes:", train_dataset.class_names)

# Limit batches for instant code execution/testing
train_dataset = train_dataset.take(5)
validation_dataset = validation_dataset.take(2)


# ------------------------------------------------------------
# PART 1: MANUAL CNN MODEL (Trained from Scratch)
# ------------------------------------------------------------
print("\n--- Training Model 1: Manual CNN from Scratch ---")

# Normalization wrapper for manual CNN (scales 0-255 to 0-1)
normalization_layer = tf.keras.layers.Rescaling(1./255)
manual_train = train_dataset.map(lambda x, y: (normalization_layer(x), y))
manual_val = validation_dataset.map(lambda x, y: (normalization_layer(x), y))

manual_cnn = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(150, 150, 3)),
    MaxPooling2D(2, 2),
    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D(2, 2),
    Flatten(),
    Dense(64, activation='relu'),
    Dropout(0.3),
    Dense(1, activation='sigmoid') # Binary classification output
])

manual_cnn.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
history_manual = manual_cnn.fit(manual_train, validation_data=manual_val, epochs=3)

# Evaluate Manual CNN
_, manual_acc = manual_cnn.evaluate(manual_val)
print(f"Manual CNN Accuracy: {manual_acc:.4f}")


# ------------------------------------------------------------
# PART 2: TRANSFER LEARNING (MobileNetV2)
# ------------------------------------------------------------
print("\n--- Training Model 2: Pre-trained MobileNetV2 ---")

# Preprocessing wrapper for MobileNetV2 (-1 to 1 scaling)
tl_train = train_dataset.map(lambda x, y: (preprocess_input(tf.cast(x, tf.float32)), y))
tl_val = validation_dataset.map(lambda x, y: (preprocess_input(tf.cast(x, tf.float32)), y))

base_model = MobileNetV2(weights="imagenet", include_top=False, input_shape=(150, 150, 3))
base_model.trainable = False

transfer_model = Sequential([
    base_model,
    GlobalAveragePooling2D(),
    Dense(64, activation="relu"),
    Dropout(0.3),
    Dense(1, activation="sigmoid")
])

transfer_model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
history_tl = transfer_model.fit(tl_train, validation_data=tl_val, epochs=3)

# Evaluate Transfer Model
_, tl_acc = transfer_model.evaluate(tl_val)
print(f"Transfer Learning Accuracy: {tl_acc:.4f}")


# ------------------------------------------------------------
# PART 3: ENSEMBLE COMBINATION (Averaging Predictions)
# ------------------------------------------------------------
print("\n--- Combining Outputs for Ensemble Accuracy ---")

# 1. Gather predictions from both models on the validation set
manual_preds = []
transfer_preds = []
true_labels = []

for images, labels in manual_val:
    # Get probability predictions (values between 0 and 1)
    p1 = manual_cnn.predict(images, verbose=0)
    manual_preds.extend(p1)

for images, labels in tl_val:
    p2 = transfer_model.predict(images, verbose=0)
    transfer_preds.extend(p2)
    true_labels.extend(labels.numpy())

manual_preds = np.array(manual_preds)
transfer_preds = np.array(transfer_preds)
true_labels = np.array(true_labels)

# 2. Average the output probabilities (Soft Voting)
ensemble_preds = (manual_preds + transfer_preds) / 2.0

# 3. Convert probabilities to binary class choices (threshold at 0.5)
ensemble_classes = (ensemble_preds >= 0.5).astype(int).flatten()

# 4. Calculate Final Ensemble Accuracy
correct_count = np.sum(ensemble_classes == true_labels)
ensemble_accuracy = correct_count / len(true_labels)

print(f"\n========================================")
print(f"Manual CNN Accuracy:       {manual_acc:.4f}")
print(f"Transfer Learning Accuracy: {tl_acc:.4f}")
print(f"Combined Ensemble Accuracy: {ensemble_accuracy:.4f}")
print(f"========================================")
```

```python
# ============================================================
# STEP 1: PRE-TRAINED MODEL (MOBILE_NET_V2) TRAINING & FINE-TUNING
# ============================================================

import tensorflow as tf
import os
import numpy as np
import matplotlib.pyplot as plt

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, GlobalAveragePooling2D
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input

# 1. Ensure dataset exists (Downloads if not already present)
if not os.path.exists("PetImages"):
    print("Downloading dataset...")
    !wget -q https://download.microsoft.com/download/3/E/1/3E1C3F21-ECDB-4869-8368-6DEBA77B919F/kagglecatsanddogs_5340.zip
    !unzip -q kagglecatsanddogs_5340.zip

    # Clean corrupted images
    for folder_name in ("Cat", "Dog"):
        folder_path = os.path.join("PetImages", folder_name)
        for fname in os.listdir(folder_path):
            fpath = os.path.join(folder_path, fname)
            try:
                with open(fpath, "rb") as fobj:
                    is_jfif = b"JFIF" in fobj.peek(10)
            except Exception:
                is_jfif = False
            if not is_jfif:
                os.remove(fpath)

# 2. Load dataset
train_dataset = tf.keras.utils.image_dataset_from_directory(
    "PetImages", validation_split=0.2, subset="training", seed=42, image_size=(150, 150), batch_size=32
)
validation_dataset = tf.keras.utils.image_dataset_from_directory(
    "PetImages", validation_split=0.2, subset="validation", seed=42, image_size=(150, 150), batch_size=32
)

print("Classes:", train_dataset.class_names)

# Limit batches for fast execution/testing (remove or adjust `.take()` for full training)
train_dataset = train_dataset.take(5)
validation_dataset = validation_dataset.take(2)

# 3. Preprocess images specifically for MobileNetV2 (-1 to 1 scaling)
tl_train = train_dataset.map(lambda x, y: (preprocess_input(tf.cast(x, tf.float32)), y))
tl_val = validation_dataset.map(lambda x, y: (preprocess_input(tf.cast(x, tf.float32)), y))

# 4. Load Pre-trained Base Model (ImageNet Weights)
base_model = MobileNetV2(weights="imagenet", include_top=False, input_shape=(150, 150, 3))
base_model.trainable = False # Freeze base features initially

# 5. Build Custom Classification Head
transfer_model = Sequential([
    base_model,
    GlobalAveragePooling2D(),
    Dense(64, activation="relu"),
    Dropout(0.3),
    Dense(1, activation="sigmoid") # Binary classification output
])

# 6. Compile and Train Initial Head
transfer_model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

print("\n--- Training Pre-trained Classification Head ---")
history_tl = transfer_model.fit(tl_train, validation_data=tl_val, epochs=3)

# 7. Fine-Tuning Phase (Unfreeze last 20 layers of MobileNetV2)
base_model.trainable = True
for layer in base_model.layers[:-20]:
    layer.trainable = False

transfer_model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.00001), # Tiny learning rate
    loss='binary_crossentropy',
    metrics=['accuracy']
)

print("\n--- Fine-Tuning Pre-trained Model ---")
history_fine = transfer_model.fit(tl_train, validation_data=tl_val, epochs=3)

# 8. Final Evaluation & Saving Predictions for Ensembling
loss, pretrained_accuracy = transfer_model.evaluate(tl_val)
print(f"\nPre-trained Model Final Accuracy: {pretrained_accuracy:.4f}")

# Extract probability predictions to use later in our ensemble step
pretrained_preds = []
pretrained_true_labels = []
for images, labels in tl_val:
    preds = transfer_model.predict(images, verbose=0)
    pretrained_preds.extend(preds)
    pretrained_true_labels.extend(labels.numpy())

pretrained_preds = np.array(pretrained_preds)
pretrained_true_labels = np.array(pretrained_true_labels)
```
